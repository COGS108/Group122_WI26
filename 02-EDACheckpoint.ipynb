### NBA Player Statistics 

This dataset includes statistics of each individual player while on a certain team from 1947-present day with the following columns:
* season — The NBA season, labeled by the year it ended
* player — The player's full name. Text.
* player_id — Basketball Reference's unique identifier for each player 
* age — The player's age in years during that season.
* team — The three-letter abbreviation for the team the player was on 
* pos — Position: PG (point guard), SG (shooting guard), SF (small forward), PF (power forward), or C (center).
* g — Games played. Count of games the player appeared in during the season, out of a maximum 82.
* gs — Games started. How many of those games the player started in the opening lineup.
* mp — Minutes played. Total minutes across the entire season.
Efficiency Metrics
* per — Player Efficiency Rating. A per-minute stat that attempts to summarize all a player's statistical contributions into one number. League average is set to 15 every season. An All-Star is typically around 20-25, and historically great seasons exceed 30.
* ts_percent — True Shooting Percentage. Measures shooting efficiency accounting for 2-pointers, 3-pointers, and free throws all together. Expressed as a proportion. League average is roughly 54-56%, and elite shooters exceed 60%.
Shot Selection
* x3p_ar — Three-Point Attempt Rate. The fraction of a player's field goal attempts that are three-pointers. A player who shoots nothing but threes would be 1.0; a center who never shoots threes would be near 0. Expressed as a proportion.
* f_tr — Free Throw Rate. Ratio of free throw attempts to field goal attempts. Measures how often a player gets to the line relative to how much they shoot. A higher number means the player draws fouls frequently. Expressed as a proportion.
Opportunity Rate Stats (the _percent columns)
These are all rate stats that estimate what share of available opportunities a player captured while on the floor, making them independent of pace, team, and lineup. They are expressed as percentages (e.g., 10.5 means 10.5%).
* orb_percent — Offensive Rebound Percentage. Estimated share of available offensive rebounds a player grabbed while on the floor.
* drb_percent — Defensive Rebound Percentage. Estimated share of available defensive rebounds a player grabbed while on the floor.
* trb_percent — Total Rebound Percentage. Estimated share of all available rebounds a player grabbed while on the floor.
* ast_percent — Assist Percentage. share of teammate field goals a player assisted on while on the floor.
* stl_percent — Steal Percentage. share of opponent possessions that ended in a steal by this player while on the floor.
* blk_percent — Block Percentage. share of opponent two-point field goal attempts that were blocked by this player while on the floor. 
* tov_percent — Turnover Percentage. share of a player's own possessions (plays where they had the ball) that ended in a turnover. 
* usg_percent — Usage Percentage. share of team possessions a player used while on the floor, where a possession is "used" by shooting, drawing a foul, or turning the ball over. League average is around 20%. Stars typically sit at 28-35%.
Win Shares
* ows — Offensive Win Shares. how many wins a player contributed through their offensive performance. Roughly, a full-season starter might contribute 1-4.
* dws — Defensive Win Shares. Same concept, for defensive performance.
* ws — Win Shares. The sum of ows and dws. Total estimated wins contributed. An average starter produces about 3-5 per season; historically great seasons can reach 15+.
* ws_48 — Win Shares per 48 Minutes. Win shares put on a per-game basis so players with different playing time can be compared. League average is around 0.100.
* obpm — Offensive Box Plus/Minus. Estimates a player's offensive contribution in points per 100 possessions relative to a league-average player, using only box score stats. 0 is league average, +5 is All-Star level.
* dbpm — Defensive Box Plus/Minus. Same concept for defense. Defense is harder to capture in box scores, so this metric is noisier than obpm.
* bpm — Box Plus/Minus. The sum of obpm and dbpm. Overall estimated impact in points per 100 possessions relative to league average. 0 is average, +5 is All-Star, +8 or above is MVP territory, negative values indicate below-average players.
* vorp — Value Over Replacement Player. Translates BPM into a cumulative total value above what a "replacement level" player would provide over the same minutes. Unlike BPM which is a rate, VORP rewards playing more minutes. A positive VORP means the player added value above replacement, 0 or negative means they were replacement-level or worse.

Some concerns with the data is that several columns have meaningful amounts of missing data. For example, x3p_ar is only available for about 26,600 of 33,140 rows, reflecting that three-point statistics weren't tracked in the early decades of the league. #### Analysis of the Distributions of Predictors
__PER (Player Efficiency Rating)__: Roughly bell-shaped with a median of 12.7, which is slightly below the known league average of 15. This is expected because the dataset includes all rostered players, bench players and the players at the end of the rotation drag the median down. The long left tail represents players who played very few minutes and had poor outcomes in that limited time.

__Win Shares__: Heavily right-skewed with a median of 1.3 and mean of 2.3, so the mean is being pulled up significantly by a small number of star players who accumulate very high win shares. Most players cluster near 0 to 2, which tells us that a large portion of the roster is made up of role players and bench contributors rather than high impact starters.

__BPM (Box Plus/Minus)__: Fairly symmetric and centered just below 0. This makes sense as BPM measures performance relative to league average, so most players sitting below 0 reflects that the average NBA roster spot is below league average when you account for garbage time minutes and fringe players. The distribution looks very normal compared to the other six.

__VORP (Value Over Replacement Player)__: Extremely skewed to the right with a massive spike at 0. The median being at 0, means more than half the players in the dataset provide essentially no value above a replacement-level player. The long right tail is driven by a handful of elite players. This is the most lopsided distribution of the group and suggests VORP may need a log transformation before being used as a predictor.

__Usage %__ : The most normally distributed looking of the six predictors, centered around a median of 18.1% and mean of 18.6%, which is close to the known league average of around 20%. The slight skew to the right reflects the small number of stars. 

__Minutes Played__: Bimodal distribution with a large spike at low minutes (0–200) representing players that barely played such as two way players or players at the end of the bench. A second hump from roughly 500–3000 for regular rotation players and starters. The mean (1094) is much higher than the median (1015) due to the high-minutes starters bringing it up.

```python
injury_r['Date'] = pd.to_datetime(injury_r['Date'])
injury_a['Date'] = pd.to_datetime(injury_a['Date'])

# Count total injuries per player
injury_counts = injury_r.groupby('Relinquished').size().reset_index(name='total_injuries')

# Calculate recovery days
merged_injury = pd.merge(
    injury_r[['Date','Relinquished','Notes']].rename(columns={'Date':'injury_date','Relinquished':'player'}),
    injury_a[['Date','Acquired']].rename(columns={'Date':'clear_date','Acquired':'player'}),
    on='player'
)
merged_injury = merged_injury[merged_injury['clear_date'] > merged_injury['injury_date']]
merged_injury = merged_injury.sort_values('clear_date').groupby(['player','injury_date']).first().reset_index()
merged_injury['days_missed'] = (merged_injury['clear_date'] - merged_injury['injury_date']).dt.days

# Total days missed per player
days_missed = merged_injury.groupby('player')['days_missed'].sum().reset_index(name='total_days_missed')

# Combine into one summary df
injury_summary = pd.merge(injury_counts, days_missed,
                           left_on='Relinquished', right_on='player', how='outer')
injury_summary = injury_summary.drop(columns='player').rename(columns={'Relinquished':'player'})

injury_summary
```

```python
# Copy of other datasets to merge with injury
import unicodedata

df_stats_injuries = df_stats.copy()
df_social_injuries = df_social.copy()
df_injury_summary = injury_summary.copy()



def remove_accents(name):
    return ''.join(
        c for c in unicodedata.normalize('NFD', name)
        if unicodedata.category(c) != 'Mn'
    )

# Apply to both datasets
df_stats_injuries['player'] = df_stats_injuries['player'].apply(remove_accents)
df_injury_summary['player'] = df_injury_summary['player'].apply(remove_accents)

df_injury_summary['player'] = df_injury_summary['player'].str.replace(r'\(.*?\)\s*', '', regex=True).str.strip()
df_injury_summary['player'] = df_injury_summary['player'].apply(
    lambda x: x.split('/')[-1].strip() if '/' in x else x
)
df_injury_summary['player'] = df_injury_summary['player'].str.replace('.', '', regex=False)

df_stats_injuries['player'] = df_stats_injuries['player'].str.replace('.', '', regex=False)

# Generalize Players
df_stats_injuries['player'] = df_stats_injuries['player'].str.lower().str.strip()
df_social_injuries['player'] = df_social_injuries['player'].str.lower().str.strip()
df_injury_summary['player'] = df_injury_summary['player'].str.lower().str.strip()



# Merge
df_socinj_eda = pd.merge(df_social_injuries, df_injury_summary, on='player', how='left')
df_socinj_eda['total_injuries'] = df_injury_summary['total_injuries'].fillna(0)
df_socinj_eda['total_days_missed'] = df_injury_summary['total_days_missed'].fillna(0)

df_stainj_eda = pd.merge(df_stats_injuries, df_injury_summary, on='player', how='left')
df_stainj_eda['total_injuries'] = df_stainj_eda['total_injuries'].fillna(0)
df_stainj_eda['total_days_missed'] = df_stainj_eda['total_days_missed'].fillna(0)

matched_stats = df_injury_summary[df_injury_summary['player'].isin(df_stats_injuries['player'])]
print(f"Injury players matched in df_stats: {matched_stats.shape[0]}")
print(f"Total injury players: {df_injury_summary.shape[0]}")

matched_stats = df_injury_summary[df_injury_summary['player'].isin(df_social_injuries['player'])]
print(f"Injury players matched in df_social: {matched_stats.shape[0]}")
print(f"Total injury players: {df_injury_summary.shape[0]}")
```
