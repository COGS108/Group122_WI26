## Background and Prior Work

Sports betting markets for player proposition bets, such as predicting how many points a player will score in a game, have expanded rapidly mainly due to the growth of legalized online wagering. Sportsbook set player point “lines” that behave as predictions of expected performance while including historical statistics, matchup info, and market demand. Since these lines already reflect modeling and large amounts of information, beating them consistently is challenging and provides a meaningful benchmark for evaluating predictive models. The goal of our project is to investigate whether a machine learning model, XGBoost, is able to predict individual player points better than sportsbook lines in games.

Prior data science work has demonstrated similar approaches. A project published onto Towards Data Science 1 used historical NBA player stats and machine learning regression models to predict the Most Valuable Player outcomes. The study represents how feature engineering and model techniques can be used to capture the patterns in each player's performance using data. Although the project honed in on the predictions based on the award, it supports the idea that regression-based machine learning models can successfully model NBA player performance.

Another data science project used XGBoost to predict a player’s points per game throught the season. This project published by Arving Rangarajan on August 28, 2025 2, used many player stats including some from two or seasons before and comparison from previous season to current season to predict with different levels of confidence the player’s performance in the season overall. Essentially this project is doing a similar process to what we have intended but in search of a different outcome. Instead of predicting a player’s performance for the entire season we want to predict their performance for a game and compare it to the sportsbook lines. What we can learn from this project is the importance of using many different variables, they used 13 variables to make their predictions. We intend to do the same but also include team stats because they have far more importance in specific games than in the entire season.

Finally, we found another project that focuses on predicting game results2using different machine learning models. This is probably the most similar to our project because it predicts outcomes of singular games instead of season overall. This is a very complex project and has a very complex explanation. An aspect of the project that stuck out to us is that XGBoost appeared to be the most accurate out of the seven machine learning models used.

Our key takeaway from all three projects is that it is possible to accurately predict NBA stats using machine learning models with the use of good data (historical & trends).
